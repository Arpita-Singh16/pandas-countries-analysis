```python
import numpy as np 
import pandas as pd
```

```python
df=pd.read_csv("Countries.csv")
```

```python
df.head()
```

### Q1. which country has the highest population

```python
df[df['population'] == df['population'].max()]['country']
```

### Q2. what is the capital of the country with highest population

```python
df[df['population'] == df['population'].max()]['capital_city']
```

### Q3. which country has the least population

```python
df[df['population'] == df['population'].min()]['country']
```

### Q4. what is the capital of the country with least population

```python
df[df['population'] == df['population'].min()]['capital_city']
```

### Q5. give me top 5 countries with highest democratic score

```python
df.sort_values(by = 'democracy_score',ascending = False, inplace = True)
```

### Q7. how many countries lie in eastern europe region

### Q8. who is the political leader of the 2nd highest populated country

```python
df[df['population'] == df['population'].nlargest(2).iloc[1]]['political_leader']
```

### Q9. how many countries are there whose political leader are unknown

```python
df[df['political_leader'].isna()]['country'].count()
```

### Q10. how many countries have republic in there full name

```python
count = 0
def counting(txt):
    global count
    if 'republic' in txt.lower():
        count += 1
    return txt
df['country_long'] = df['country_long'].apply(counting)
print(count)
```
